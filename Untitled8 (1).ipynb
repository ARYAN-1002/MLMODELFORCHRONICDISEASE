**INITIAL MAIN PROGRAM**

```python
import yaml
from joblib import dump, load
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
class DiseasePrediction:
    # Initialize and Load the Config File
    def __init__(self, model_name=None):
        self.config = {
            'verbose': True,
            'dataset': {
                'training_data_path': '/content/train_data.csv',  # Ensure the CSV is uploaded to Google Colab
                'test_data_path': '/content/test_data.csv'
            },
            'model_save_path': '/content/',  # Save models in Colab's default directory
            'random_state': 42,
            'model': {
                'decision_tree': {'criterion': 'gini'},
                'random_forest': {'n_estimators': 100},
                'gradient_boost': {'n_estimators': 100, 'criterion': 'friedman_mse'}
            }
        }

        # Verbose
        self.verbose = self.config['verbose']
        # Load Training Data
        self.train_features, self.train_labels, self.train_df = self._load_train_dataset()
        # Load Test Data
        self.test_features, self.test_labels, self.test_df = self._load_test_dataset()
        # Feature Correlation in Training Data
        self._feature_correlation(data_frame=self.train_df, show_fig=False)
        # Model Definition
        self.model_name = model_name
        # Model Save Path
        self.model_save_path = self.config['model_save_path']

    # Function to Load Train Dataset
    def _load_train_dataset(self):
        df_train = pd.read_csv(self.config['dataset']['training_data_path'])
        cols = df_train.columns
        cols = cols[:-2]
        train_features = df_train[cols]
        train_labels = df_train['prognosis']

        if self.verbose:
            print("Length of Training Data: ", df_train.shape)
            print("Training Features: ", train_features.shape)
            print("Training Labels: ", train_labels.shape)
        return train_features, train_labels, df_train

    # Function to Load Test Dataset
    def _load_test_dataset(self):
        df_test = pd.read_csv(self.config['dataset']['test_data_path'])
        cols = df_test.columns
        cols = cols[:-1]
        test_features = df_test[cols]
        test_labels = df_test['prognosis']

        if self.verbose:
            print("Length of Test Data: ", df_test.shape)
            print("Test Features: ", test_features.shape)
            print("Test Labels: ", test_labels.shape)
        return test_features, test_labels, df_test

    # Features Correlation
    def _feature_correlation(self, data_frame=None, show_fig=False):
        numeric_data = data_frame.select_dtypes(include=[np.number])
        corr = numeric_data.corr()

        sn.heatmap(corr, square=True, annot=False, cmap="YlGnBu")
        plt.title("Feature Correlation")
        plt.tight_layout()
        if show_fig:
            plt.show()
        plt.savefig('/content/feature_correlation.png')

    # Model Selection
    def select_model(self):
        if self.model_name == 'mnb':
            self.clf = MultinomialNB()
        elif self.model_name == 'decision_tree':
            self.clf = DecisionTreeClassifier(criterion=self.config['model']['decision_tree']['criterion'])
        elif self.model_name == 'random_forest':
            self.clf = RandomForestClassifier(n_estimators=self.config['model']['random_forest']['n_estimators'])
        elif self.model_name == 'gradient_boost':
            self.clf = GradientBoostingClassifier(n_estimators=self.config['model']['gradient_boost']['n_estimators'],
                                                  criterion=self.config['model']['gradient_boost']['criterion'])
        return self.clf

    # ML Model
    def train_model(self):
        # Train on the full training dataset
        classifier = self.select_model()
        classifier = classifier.fit(self.train_features, self.train_labels)

        # Training Accuracy
        train_pred = classifier.predict(self.train_features)
        training_accuracy = accuracy_score(self.train_labels, train_pred)

        # Trained Model Evaluation on Test Dataset
        test_pred = classifier.predict(self.test_features)
        test_accuracy = accuracy_score(self.test_labels, test_pred)
        conf_mat = confusion_matrix(self.test_labels, test_pred)
        clf_report = classification_report(self.test_labels, test_pred)

        if self.verbose:
            print('\nTraining Accuracy: ', training_accuracy)
            print('\nTest Data Predictions: ', test_pred)
            print('\nTest Accuracy: ', test_accuracy)
            print('\nTest Confusion Matrix: \n', conf_mat)
            print('\nClassification Report: \n', clf_report)

        # Plot Feature Importances
        if hasattr(classifier, 'feature_importances_'):
            self._plot_feature_importances(classifier)

        # Save Trained Model
        dump(classifier, str(self.model_save_path + self.model_name + ".joblib"))

    # Function to Plot Feature Importances
    def _plot_feature_importances(self, classifier):
        feature_importances = classifier.feature_importances_
        features = self.train_features.columns

        # Create a DataFrame for plotting
        importance_df = pd.DataFrame({
            'Feature': features,
            'Importance': feature_importances
        })

        # Sort the DataFrame
        importance_df = importance_df.sort_values(by='Importance', ascending=False)

        # Plot
        plt.figure(figsize=(10, 8))
        sn.barplot(x='Importance', y='Feature', data=importance_df)
        plt.title(f"Feature Importances for {self.model_name}")
        plt.tight_layout()
        plt.savefig(f'/content/{self.model_name}_feature_importances.png')
        plt.show()

    # Function to Make Predictions on Test Data
    def make_prediction(self, saved_model_name=None, test_data=None):
        try:
            # Load Trained Model
            clf = load(str(self.model_save_path + saved_model_name + ".joblib"))
        except Exception as e:
            print("Model not found...")

        if test_data is not None:
            result = clf.predict(test_data)
            return result
        else:
            result = clf.predict(self.test_features)
        accuracy = accuracy_score(self.test_labels, result)
        clf_report = classification_report(self.test_labels, result)
        return accuracy, clf_report
```

```python
if __name__ == "__main__":
    # Model Currently Training
    current_model_name = 'random_forest'  # Choose the model that supports feature importance
    # Instantiate the Class
    dp = DiseasePrediction(model_name=current_model_name)
    # Train the Model
    dp.train_model()
    # Get Model Performance on Test Data
    test_accuracy, classification_report = dp.make_prediction(saved_model_name=current_model_name)
    print("Model Test Accuracy: ", test_accuracy)
    print("Test Data Classification Report: \n", classification_report)
```

**PROGRAM FOR FEATURE IMPORTANCE OF EACH OUTPUT CLASS CONSIDERING ALL 132 FEATURES**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

class DiseasePrediction:
    def __init__(self):
        self.config = {
            'dataset': {
                'training_data_path': '/content/train_data.csv'
            }
        }
        # Load Training Data
        self.train_features, self.train_labels, self.train_df = self._load_train_dataset()

    def _load_train_dataset(self):
        df_train = pd.read_csv(self.config['dataset']['training_data_path'])
        cols = df_train.columns
        cols = cols[:-2]
        train_features = df_train[cols]
        train_labels = df_train['prognosis']
        return train_features, train_labels, df_train

    def calculate_feature_importance(self):
        # Unique classes (diseases) from the dataset
        unique_classes = self.train_labels.unique()
        feature_importances = {}

        for disease in unique_classes:
            # Create a binary target variable for each class
            y = (self.train_labels == disease).astype(int)

            # Train a RandomForest model
            clf = RandomForestClassifier(n_estimators=100, random_state=42)
            clf.fit(self.train_features, y)

            # Extract feature importances
            importances = clf.feature_importances_
            feature_importances[disease] = importances

            # Plot feature importances for the current disease
            self.plot_feature_importance(disease, importances)

        return feature_importances

    def plot_feature_importance(self, disease, importances):
        # Get feature names
        features = self.train_features.columns

        # Sort feature importances in descending order
        indices = np.argsort(importances)[::-1]

        # Plot
        plt.figure(figsize=(12, 8))
        plt.title(f"Feature Importance for {disease}")
        plt.bar(range(len(indices)), importances[indices], align="center")
        plt.xticks(range(len(indices)), features[indices], rotation=90)
        plt.xlabel("Feature")
        plt.ylabel("Importance")
        plt.tight_layout()
        plt.savefig(f'/content/feature_importance_{disease}.png')
        plt.show()

# Usage in Jupyter Notebook
if __name__ == "__main__":
    # Instantiate the Class
    dp = DiseasePrediction()
    # Calculate and plot feature importance for each disease
    dp.calculate_feature_importance()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from joblib import dump

class DiseasePrediction:
    def __init__(self, model_name=None, feature_importance_threshold=0.01):
        self.config = {
            'verbose': True,
            'dataset': {
                'training_data_path': '/content/train_data.csv',
                'test_data_path': '/content/test_data.csv'
            },
            'model_save_path': '/content/',
            'random_state': 42,
            'model': {
                'decision_tree': {'criterion': 'gini'},
                'random_forest': {'n_estimators': 100},
                'multinomial_nb': {},
                'gradient_boost': {'n_estimators': 100, 'criterion': 'friedman_mse'}
            }
        }
        # Feature importance threshold
        self.feature_importance_threshold = feature_importance_threshold
        # Load Training Data
        self.train_features, self.train_labels, self.train_df = self._load_train_dataset()
        # Load Test Data
        self.test_features, self.test_labels, self.test_df = self._load_test_dataset()
        self.model_name = model_name

    def _load_train_dataset(self):
        df_train = pd.read_csv(self.config['dataset']['training_data_path'])
        cols = df_train.columns[:-2]
        train_features = df_train[cols]
        train_labels = df_train['prognosis']
        return train_features, train_labels, df_train

    def _load_test_dataset(self):
        df_test = pd.read_csv(self.config['dataset']['test_data_path'])
        cols = df_test.columns[:-1]
        test_features = df_test[cols]
        test_labels = df_test['prognosis']
        return test_features, test_labels, df_test

    def calculate_feature_importance(self):
        unique_classes = self.train_labels.unique()
        feature_importances = {}

        for disease in unique_classes:
            y = (self.train_labels == disease).astype(int)
            clf = RandomForestClassifier(n_estimators=100, random_state=42)
            clf.fit(self.train_features, y)
            importances = clf.feature_importances_
            feature_names = self.train_features.columns

            # Select important features based on the threshold
            important_features = [feature_names[i] for i in range(len(importances)) if importances[i] > self.feature_importance_threshold]
            important_importances = [importances[i] for i in range(len(importances)) if importances[i] > self.feature_importance_threshold]

            feature_importances[disease] = important_features

            # Plot and print the important features for the current disease
            self.plot_feature_importance(disease, important_features, important_importances)
            print(f"Important Features for {disease}: {important_features}")

        return feature_importances

    def plot_feature_importance(self, disease, features, importances):
        # Sort feature importances in descending order
        indices = np.argsort(importances)[::-1]

        # Plot
        plt.figure(figsize=(12, 8))
        plt.title(f"Feature Importance for {disease}")
        plt.bar(range(len(indices)), np.array(importances)[indices], align="center")
        plt.xticks(range(len(indices)), np.array(features)[indices], rotation=90)
        plt.xlabel("Feature")
        plt.ylabel("Importance")
        plt.tight_layout()
        plt.savefig(f'/content/feature_importance_{disease}.png')
        plt.show()

    def select_important_features(self, feature_importances):
        # Select the union of important features across all classes
        selected_features = set()
        for disease, features in feature_importances.items():
            selected_features.update(features)

        selected_features = list(selected_features)
        # Filter the training and test data by selected features
        self.train_features = self.train_features[selected_features]
        self.test_features = self.test_features[selected_features]

    # Model selection based on user input
    def select_model(self):
        if self.model_name == 'random_forest':
            return RandomForestClassifier(n_estimators=self.config['model']['random_forest']['n_estimators'], random_state=self.config['random_state'])
        elif self.model_name == 'decision_tree':
            return DecisionTreeClassifier(criterion=self.config['model']['decision_tree']['criterion'], random_state=self.config['random_state'])
        elif self.model_name == 'multinomial_nb':
            return MultinomialNB()
        elif self.model_name == 'gradient_boost':
            return GradientBoostingClassifier(n_estimators=self.config['model']['gradient_boost']['n_estimators'],
                                              criterion=self.config['model']['gradient_boost']['criterion'],
                                              random_state=self.config['random_state'])
        else:
            raise ValueError(f"Model {self.model_name} is not supported!")

    def retrain_model_on_important_features(self):
        # Select the model for training based on user choice
        classifier = self.select_model()
        classifier.fit(self.train_features, self.train_labels)

        # Make predictions on the training data to calculate training accuracy
        train_pred = classifier.predict(self.train_features)
        train_accuracy = accuracy_score(self.train_labels, train_pred)

        # Make predictions on the test data to calculate test accuracy
        test_pred = classifier.predict(self.test_features)
        test_accuracy = accuracy_score(self.test_labels, test_pred)

        clf_report = classification_report(self.test_labels, test_pred)
        conf_mat = confusion_matrix(self.test_labels, test_pred)
        print(f'\nCurrent Model : {self.model_name}')
        print(f'Training Accuracy: {train_accuracy}')
        print(f'Test Accuracy: {test_accuracy}')
        print(f'Confusion Matrix:\n{conf_mat}')
        print(f'Classification Report:\n{clf_report}')

        # Save the model
        dump(classifier, f"{self.config['model_save_path']}{self.model_name}_important_features.joblib")


if __name__ == "__main__":
    # Model Currently Training (Choose between 'random_forest', 'decision_tree', 'multinomial_nb', or 'gradient_boost')
    current_model_name = 'multinomial_nb'  # Change this to any supported model

    # Instantiate the Class
    dp = DiseasePrediction(model_name=current_model_name, feature_importance_threshold=0.01)

    # Step 1: Calculate Feature Importance
    feature_importances = dp.calculate_feature_importance()

    # Step 2: Select Important Features
    dp.select_important_features(feature_importances)

    # Step 3: Retrain the Model on Important Features and print Training Accuracy
    dp.retrain_model_on_important_features()
```
